```python
# Collect ALL images (labelled or not) + any matching YOLO txt labels into
# datasett/images and datasett/labels, zip the datasett folder, and download it.

import os
import shutil
from pathlib import Path

# ── CONFIG ────────────────────────────────────────────────────────────────────
SOURCE_IMAGES_DIR = Path("unlabelled_val_random")          # all images run through predict
RUN_LABELS_DIR     = Path("/content/runs/pose/predict/labels")   # where YOLO saved save_txt=True output
DATASETT_DIR       = Path("datasett_2")
IMAGES_OUT         = DATASETT_DIR / "images"
LABELS_OUT         = DATASETT_DIR / "labels"
IMAGE_EXTS         = {".png", ".jpg", ".jpeg"}
ZIP_NAME           = "datasett_2"  # produces datasett.zip

# ── COLLECT ───────────────────────────────────────────────────────────────────
os.makedirs(IMAGES_OUT, exist_ok=True)
os.makedirs(LABELS_OUT, exist_ok=True)

source_images = sorted(
    f for f in SOURCE_IMAGES_DIR.iterdir()
    if f.suffix.lower() in IMAGE_EXTS
)
print(f"Found {len(source_images)} source images in {SOURCE_IMAGES_DIR}")

copied_images = 0
copied_labels = 0
no_detection  = []

for image_file in source_images:
    # copy the image regardless of whether it has a detection
    shutil.copy2(image_file, IMAGES_OUT / image_file.name)
    copied_images += 1

    # copy matching label file if the model produced one
    label_file = RUN_LABELS_DIR / f"{image_file.stem}.txt"
    if label_file.exists():
        shutil.copy2(label_file, LABELS_OUT / label_file.name)
        copied_labels += 1
    else:
        no_detection.append(image_file.name)

print(f"Copied {copied_images} images  -> {IMAGES_OUT}")
print(f"Copied {copied_labels} labels  -> {LABELS_OUT}")

if no_detection:
    print(f"\n{len(no_detection)} images had no detections (no label file):")
    for n in no_detection:
        print(f"  {n}")

# ── ZIP ───────────────────────────────────────────────────────────────────────
zip_path = shutil.make_archive(ZIP_NAME, "zip", DATASETT_DIR)
print(f"\nZipped to: {zip_path}")
```
